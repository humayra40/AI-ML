## Question 1: Filter Numbers Divisible by 4 but Not by 8

**Problem**

Given a list of integers, extract all numbers that are divisible by 4 but not divisible by 8.

**Input format:** Space-separated integers.

**Output format:** A list of numbers satisfying the condition.

**Example input**
```text
4 6 8 12 16 20 24 28
```

**Expected output**
```text
[4, 12, 20, 28]
```

**How the code works:** The loop converts each input to an integer. It adds a number to `result` only when the remainder after division by 4 is zero and the remainder after division by 8 is not zero.

### Solution

````python
numbers = input().split()
result = []

for i in numbers:
  number = int(i)
  if number%4 == 0 and number%8 !=0:
    result.append(number)

print(result)
````

## Question 2: Calculate a Ratio with Exception Handling

**Problem**

Write a function `compute_ratio(x, y)` that returns `(x + y) / (x * y)`. Use exception handling to manage division by zero.

**Input format:** Two values. For this solution, enter two whole numbers on one line, separated by a space.

**Output format:** Computed result OR error message.

**Example input**
```text
5 0
```

**Expected output**
```text
Error: Division not possible
```

**How the code works:** The function divides the sum of the two numbers by their product. If either number is zero, `except ZeroDivisionError` returns the error message. For a nonzero example, input `4 2` gives `0.75`.

### Solution

````python
def compute_ratio(x,y):
  try:
    return(x+y)/(x*y)
  except ZeroDivisionError:
    return "Error: Division not possible"

numbers = input().split()
x = int(numbers[0])
y = int(numbers[1])


result = compute_ratio(x,y)
print(result)
````

## Question 3: Find the Longest Word in a Text File

**Problem**

Write a Python program that reads a text file and finds the longest word.

**Requirements**

- Open and read data from a file (for example, `data.txt`).
- If multiple words have the same longest length, print any one of them.

**Input format:** A text file containing multiple lines of text.

**Output format:** The longest word.

**Example file content (`data.txt`)**
```text
Python is a wonderful language for machine learning.
```

**Expected output**
```text
wonderful
```

**File setup:** The code uses `./sample_data/data.txt`. Create a folder named `sample_data` in the notebook's working directory, then save the example sentence above in a file named `data.txt` inside that folder. In Colab, upload the file into the `sample_data` folder. The text file must contain at least one word.

**How the code works:** The code reads the file, splits its text into words, and compares their lengths. It keeps the first word with the greatest length. Punctuation counts as part of a word, so `learning.` and `wonderful` both have nine characters in this example; `wonderful` appears first.

### Solution

````python
with open("./sample_data/data.txt", "r") as file:
  content = file.read()
words = content.split()
longest = words[0]

for i in words:
  if len(i)>len(longest):
     longest = i

print(longest)
````

## Question 4: Select Words That Start with a Vowel

**Problem**

Write a program that converts a sentence into a list of words but only includes words that start with a vowel (a, e, i, o, u), ignoring case.

**Input format:** A single line sentence.

**Output format:** A list of strings.

**Example input**
```text
An apple a day keeps the doctor away
```

**Expected output**
```text
['An', 'apple', 'a', 'away']
```

**How the code works:** The program splits the sentence into words and checks the first character of each word in lowercase. Matching words are stored with their original spelling and order. A word starting with punctuation is not selected.

### Solution

````python
words = input().split()
result = []

for i in words:
  first = i[0].lower()

  if first in["a","e","i","o","u"]:
    result.append(i)

print(result)
````

## Question 5: Check Whether a Word Is a Palindrome

**Problem**

A word is called a Palindrome if reversing its letters results in the same word (ignoring case). Write a program to check whether a given string is a palindrome.

**Input format:** A single word string.

**Output format:** `Palindrome` or `Not Palindrome`.

**Example input**
```text
Radar
```

**Expected output**
```text
Palindrome
```

**How the code works:** The program converts the input to lowercase and reverses it using `[::-1]`. If the two strings match, it prints `Palindrome`; otherwise, it prints `Not Palindrome`. Spaces and punctuation are included in the comparison.

### Solution

````python
word = input()
word = word.lower()

reverse_word = word[::-1]

if word == reverse_word:
  print("Palindrome")
else:
  print("Not Palindrome")
````
